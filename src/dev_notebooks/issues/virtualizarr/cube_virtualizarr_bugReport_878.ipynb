## Bug report #878 for virtualizarr package

### Failure to read HDF5 data with data arrays of dtype='<U1'

https://github.com/zarr-developers/VirtualiZarr/issues/878

```python
%load_ext autoreload
%autoreload 2

import warnings
warnings.filterwarnings('ignore')

# Enable output of all statements in the cell
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
```

```python
import obstore
import virtualizarr as vz
from virtualizarr.parsers import HDFParser
from obspec_utils.registry import ObjectStoreRegistry
```

```python
bucket = 's3://its-live-data'
granule = 'velocity_image_pair/landsatOLI/v02/S80W170/LC08_L1GT_020121_20231013_20231102_02_T2_X_LC09_L1GT_020121_20231106_20231106_02_T2_G0120V02_P084.nc'
```

```python
# Problem exists if include '<U1' data variables with missing _FillValue: "mapping"

store = obstore.store.from_url(bucket, region="us-west-2", skip_signature=True)
registry = ObjectStoreRegistry({bucket: store})
parser = HDFParser(drop_variables=['img_pair_info'])

vds = vz.open_virtual_dataset(
    url=f"{bucket}/{granule}",
    parser=parser,
    registry=registry,
    loadable_variables=[],
    # drop_variables=['img_pair_info', 'mapping']
)
```

```python
# Problem does not exist if exclude '<U1' data variables with missing _FillValue from the parser

store = obstore.store.from_url(bucket, region="us-west-2", skip_signature=True)
registry = ObjectStoreRegistry({bucket: store})
parser = HDFParser(drop_variables=['img_pair_info', 'mapping'])

vds = vz.open_virtual_dataset(
    url=f"{bucket}/{granule}",
    parser=parser,
    registry=registry,
    loadable_variables=[],
    # drop_variables=['img_pair_info', 'mapping']
)
```
